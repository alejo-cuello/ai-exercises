# Import libraries

```python
import tiktoken
from dotenv import load_dotenv
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

load_dotenv()
```

# Create a Document for this example

```python
doc = Document(
    page_content="This is the core text content of the document.",
    metadata={"source": "user_input", "type": "tutorial"}
)
```

# Use the parameter length_function of RecursiveCharacterTextSplitter 

## String length

```python
len(doc.page_content)
```

## Default length_function

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=3,
    chunk_overlap=0
)

chunks = splitter.split_documents([doc])

print(f"Chunks quantity: {len(chunks)}")
print("Examples of chunks:")
print([c.page_content for c in chunks[0:4]])
```

## length_function = encoder

```python
tokenizer = tiktoken.get_encoding("o200k_base") # Common OpenAI Models Using this encoding name: gpt-4o, gpt-4o-mini, gpt-5.1

# Custom function to calculate actual OpenAI tokens
def count_tokens(text_string: str) -> int:
    return len(tokenizer.encode(text_string))
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=3,
    chunk_overlap=0,
    length_function=count_tokens
)

chunks = splitter.split_documents([doc])

print(f"Chunks quantity: {len(chunks)}")
print("Examples of chunks:")
print([c.page_content for c in chunks[0:4]])
```
