# Get original files

In this exercise, I'll use only the chapter 2 of the book

```python
# Load all PDFs from data folder
all_docs = []
data_dir = os.path.join("data", "raw")
files = sorted([f for f in os.listdir(data_dir) if f.endswith(".pdf")])

for file in files:
    # Extract chapter number from filename (e.g., book-chapter-00.pdf -> 00)
    match = re.search(r"book_chapter_(\d+)\.pdf", file)
    if match:
        chapter_num = match.group(1)
        loader = PyMuPDFLoader(file_path=os.path.join(data_dir, file))
        docs = loader.lazy_load()
        # Add metadata to each document (page)
        for doc in docs:
            doc.metadata.update({
                "volume": 1,
                "chapter": int(chapter_num)
            })
            all_docs.append(doc)

print(f'Length all_docs: {len(all_docs)} \n')
print(f'all_docs[0]: \n{all_docs[0]}')
```

```python
# Split into smaller chunks

chunks = []

import re

# TODO: improve the loading of figures and formulas
for doc in all_docs:
    # Access the string content specifically to avoid the AttributeError

    # 1. Sub-split long sections into chunks
    child_splitter = RecursiveCharacterTextSplitter(
        chunk_size=800, 
        chunk_overlap=80,
        separators=["\n\n", "\n", ". "]
    )

    # Use split_documents to handle the list of section Documents
    final_chunks = child_splitter.split_documents([doc]) # 
    chunks.extend(final_chunks)
```

## Save only useful metadata 

```python
keys_to_keep =["creationdate", "source", "total_pages", "page", "volume", "chapter", "start_index"]

for chunk in chunks:
    # Filter the dictionary
    chunk.metadata = {k: v for k, v in chunk.metadata.items() if k in keys_to_keep}
```

# Create tags to apply to the chunks

```python
class Tags(BaseModel):
    content_type: str = Field(
        description="The functional nature of the text fragment. The possible values must be: 'Narrative', 'Summary', 'Historical Notes', 'Examples', 'Exercises', 'Table of Contents', 'Algorithm Description'",
        default="Narrative"
    )
    contains_math_latex: bool = Field(
        default=False,
        description="Whether the chunk contains formal mathematical notation, probability equations, or LaTeX-style formulas.",
    )
    contains_regex: bool = Field(
        default=False,
        description="Whether the chunk includes regular expression patterns or explains regex syntax.",
    )
    contains_code_or_cli: bool = Field(
        default=False,
        description="Whether the text includes Python code, Unix commands (tr, sort, uniq), or pseudocode for algorithms.",
    )
    contains_table: bool = Field(
        default=False,
        description="Whether the fragment includes a data table, such as word counts, probabilities, or ASCII/Unicode mappings.",
    )
    talks_about_tokenization: bool = Field(
        default=False,
        description="Whether the chunk discusses the process of segmenting running input text into tokens[cite: 377, 1268].",
    )
    talks_about_ngrams: bool = Field(
        default=False,
        description="Whether the chunk discusses sequences of $n$ words or Markov-based models that look $n-1$ words into the past[cite: 1355, 1407, 1967].",
    )
    talks_about_language_modeling: bool = Field(
        default=False,
        description="Whether the chunk discusses the general machine learning task of predicting upcoming words or assigning probabilities to sequences[cite: 1335, 1965].",
    )
    talks_about_morphology: bool = Field(
        default=False,
        description="Whether the chunk discusses morphemes as the minimal meaning-bearing units of language[cite: 191, 192, 1271].",
    )
    linguistic_focus: str = Field(
        description="Specific languages or language families discussed as examples (e.g., Chinese, English, Spanish, Vietnamese).",
        default="General"
    )
    importance_score: str = Field(
        description="The pedagogical value of the chunk for understanding core NLP foundations. The possible values must be: 'High', 'Medium', 'Low'",
        default="Medium"
    )


pprint(Tags.model_json_schema())
```

Create the tagger to apply to the chunks:

```python
tagger = llm.with_structured_output(Tags)
```

Example with a string:

```python
result = tagger.invoke("the language model is very important to build strong systemns")
pprint(result.model_dump_json())
```

Merge tags and metadata

```python
tagged_output_with_tags_merged_with_metadata = []
for item in tagged_output:
    merged_item = {
        "index": item["index"],
        "metadata": (item["metadata"] | json.loads(item["tags"])),
        "content": item["content"]
    }
    tagged_output_with_tags_merged_with_metadata.append(Document(
        page_content=merged_item["content"],
        metadata=merged_item["metadata"],
        id=merged_item["index"]
    ))
```

## Index the tagged documents 

```python
collection_name = "langchain_docs_index"
namespace = f"chroma/{collection_name}"
embedding = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", google_api_key=api_key, task_type="RETRIEVAL_DOCUMENT")

vectorstore = Chroma(embedding_function=embedding, collection_name=collection_name, persist_directory="./data/vectors/chroma_db")

record_manager = SQLRecordManager(
    namespace=namespace,
    db_url="sqlite:///data/vectors/records.db",
)
record_manager.create_schema()
```

```python
index(
    docs_source=tagged_output_with_tags_merged_with_metadata,
    record_manager=record_manager,
    vector_store=vectorstore,
    cleanup="incremental",
    source_id_key="source",
)
```

## Retriever

```python
metadata_field_info = [
    AttributeInfo(
        name="content_type",
        description="The functional nature of the text fragment. The possible values must be: 'Narrative', 'Summary', 'Historical Notes', 'Examples', 'Exercises', 'Table of Contents', 'Algorithm Description'",
        type="str"
    ),
    AttributeInfo(
        name="contains_math_latex",
        description="Whether the chunk contains formal mathematical notation, probability equations, or LaTeX-style formulas.",
        type="bool"
    ),
    AttributeInfo(
        name="contains_regex",
        description="Whether the chunk includes regular expression patterns or explains regex syntax.",
        type="bool"
    ),
    AttributeInfo(
        name="contains_code_or_cli",
        description="Whether the text includes Python code, Unix commands (tr, sort, uniq, or pseudocode for algorithms.",
        type="bool"
    ),
    AttributeInfo(
        name="contains_table",
        description="Whether the fragment includes a data table, such as word counts, probabilities, or ASCII/Unicode mappings.",
        type="bool"
    ),
    AttributeInfo(
        name="talks_about_tokenization",
        description="Whether the chunk discusses the process of segmenting running input text into tokens[cite: 377, 1268].",
        type="bool"
    ),
    AttributeInfo(
        name="talks_about_ngrams",
        description="Whether the chunk discusses sequences of $n$ words or Markov-based models that look $n-1$ words into the past[cite: 1355, 1407, 1967].",
        type="bool"
    ),
    AttributeInfo(
        name="talks_about_language_modeling",
        description="Whether the chunk discusses the general machine learning task of predicting upcoming words or assigning probabilities to sequences[cite: 1335, 1965].",
        type="bool"
    ),
    AttributeInfo(
        name="talks_about_morphology",
        description="Whether the chunk discusses morphemes as the minimal meaning-bearing units of language[cite: 191, 192, 1271].",
        type="bool"
    ),
    AttributeInfo(
        name="linguistic_focus",
        description="Specific languages or language families discussed as examples (e.g., Chinese, English, Spanish, Vietnamese.",
        type="str"
    ),
    AttributeInfo(
        name="importance_score",
        description="The pedagogical value of the chunk for understanding core NLP foundations. The possible values must be: 'High', 'Medium', 'Low'",
        type="str"
    )
]

document_content_description = "Book about Large Language Models"
```

```python
retriever = SelfQueryRetriever.from_llm(
    llm=llm,
    vectorstore=vectorstore,
    document_contents=document_content_description,
    metadata_field_info=metadata_field_info,
    enable_limit=True,
    verbose=True,
)
```
