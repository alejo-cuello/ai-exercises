# Get original files and split into small chunks

**Important:**

- The files format must be .pdf.
- The name of the files must be: 'book_chapter_(chapter).pdf

```python
PyMuPDFLoader()
```

```python
# Load all PDFs from data folder
all_docs = []
data_dir = os.path.join("data", "raw")
files = sorted([f for f in os.listdir(data_dir) if f.endswith(".pdf")])

for file in files:
    # Extract chapter number from filename (e.g., book-chapter-00.pdf -> 00)
    match = re.search(r"book_chapter_(\d+)\.pdf", file)
    if match:
        chapter_num = match.group(1)
        loader = PyMuPDFLoader(file_path=os.path.join(data_dir, file))
        docs = loader.lazy_load()
        # Add metadata to each document (page)
        for doc in docs:
            doc.metadata.update({
                "volume": 1,
                "chapter": int(chapter_num)
            })
            all_docs.append(doc)

print(f'Length all_docs: {len(all_docs)} \n')
print(f'all_docs[0]: \n{all_docs[0]}')
```

```python
# Split into smaller chunks
# Define the headers to split on based on the Jurafsky & Martin text
headers_to_split_on = [
    ("CHAPTER", "Chapter"),    # Captures major chapters like 'Words and Tokens'
    ("2.", "Section"),         # Captures sections like 'Words' or 'Unicode'
]

chunks = []

import re

def clean_marginalia(text):
    # Fix common joined-word artifacts found in this specific text
    # Examples: "actsclitic" -> "acts", "ortokenization" -> "or"
    text = text.replace("actsclitic", "acts")
    text = text.replace("ortokenization", "or")
    
    # Generic fix: Identify if a known keyword is 'glued' to the end of a word
    keywords = ["clitic", "tokenization", "morpheme", "BPE"]
    for word in keywords:
        # Regex to find a word glued to a keyword and separate them
        text = re.sub(rf'(\w+){word}', r'\1', text)
    
    return text

# TODO: improve the loading of figures and formulas
for doc in all_docs:
    # Access the string content specifically to avoid the AttributeError
    content = clean_marginalia(doc.page_content) # Clean the text first
    # content = doc.page_content
    
    # 1. Split by logical structure
    header_splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
    sections = header_splitter.split_text(content)

    # 2. Re-attach original metadata (chapter/volume) to the new sections
    for section in sections:
        section.metadata.update(doc.metadata)

    # 3. Sub-split long sections into chunks
    child_splitter = RecursiveCharacterTextSplitter(
        chunk_size=800, 
        chunk_overlap=80,
        separators=["\n\n", "\n", ". "]
    )

    # Use split_documents to handle the list of section Documents
    final_chunks = child_splitter.split_documents(sections) # 
    chunks.extend(final_chunks)
```

```python
print(f"Split into {len(chunks)} chunks")
print("keys:")
print(vars(chunks[0]).keys())  # First chunk
print("metadata sample (first chunk):") 
print(chunks[0].metadata)
print("metadata sample (last chunk):")
print(chunks[-1].metadata)
print("page_content sample:")
print(chunks[100].page_content[:200])
```

## Keep only useful metadata 

```python
keys_to_keep =["creationdate", "source", "total_pages", "page", "volume", "chapter", "start_index"]

for chunk in chunks:
    # Filter the dictionary
    chunk.metadata = {k: v for k, v in chunk.metadata.items() if k in keys_to_keep}
```

## Analyze the chunks

Group chunks by chapter

```python
def group_docs_by_metadata(documents, key="source"):
    # Sort is required for groupby to work correctly
    sorted_docs = sorted(documents, key=lambda x: x.metadata.get(key, "unknown"))
    
    grouped = {
        k: list(g) 
        for k, g in groupby(sorted_docs, key=lambda x: x.metadata.get(key, "unknown"))
    }
    return grouped
```

```python
chunks_by_chapter = group_docs_by_metadata(chunks, "chapter")
for chapter in chunks_by_chapter:
    print(f"Chunks of the chapter {chapter}: {len(chunks_by_chapter[chapter])}")
```

## Index the chunks

```python
api_key = os.getenv("API_KEY")
embedding = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", api_key=api_key)
```

```python
collection_name = "langchain_docs_index"

get_vectorstore = partial(Chroma, embedding_function=embedding)
vector_store = get_vectorstore(
    collection_name=collection_name,
)

namespace = f"chroma/{collection_name}"
record_manager = SQLRecordManager(
    namespace=namespace,
    db_url="sqlite:///:memory:",
)
record_manager.create_schema()
```

Test the indexing with only 3 chunks

```python
index(
    docs_source=chunks[0:3],
    record_manager=record_manager,
    vector_store=vector_store,
    source_id_key="source",
    cleanup="incremental",
)
```
